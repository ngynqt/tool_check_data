```python
test_data = pd.read_csv("/home/tuong/Job/data /combined_dataset.csv")
test_data.head()
```

```python
#tách Tên KH và Facebook và SĐT vào một dataframe mới
df = test_data[['Tên KH', 'Facebook', 'SĐT', 'SĐT 2']]
df.head()
```

```python
#đổi kiểu dữ liệu của cột SĐT và SĐT 2 thành chuỗi và thêm 0 vào đầu số điện thoại nếu nó không có 0 ở đầu
df['SĐT'] = df['SĐT'].astype(str).str.replace(r'\D', '', regex=True)
df['SĐT'] = df['SĐT'].str.pad(width=10, side='left', fillchar='0')
df['SĐT 2'] = df['SĐT 2'].astype(str).str.replace(r'\D', '', regex=True)
df['SĐT 2'] = df['SĐT 2'].str.pad(width=10, side='left', fillchar='0')
df.info()
```
